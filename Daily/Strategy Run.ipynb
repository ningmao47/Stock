```python
from FinMind.data import DataLoader
import pandas as pd
import numpy as np
import talib
from datetime import datetime
import matplotlib.pyplot as plt
import os

api = DataLoader()
api.login_by_token(api_token="")
# api.login_by_token(api_token="你的API_TOKEN")

current_time = datetime.now().strftime("%Y-%m-%d")
end_date = current_time
start_date = "2020-01-01"
end_date = end_date
print(end_date)

# stock_id="6257"
# df = api.taiwan_stock_daily(
#     stock_id=stock_id,
#     start_date=start_date,
#     end_date=end_date)

# df.to_csv(f"{stock_id}.csv", index=False)

stock_list = ["2327", "2492", "3026", "3090",
              "1519", "1513", "1503", "1514",
              "2382", "2303", "2317", "2327",
              "2344", "2337", "6770", "2408", "3006",
              "3491", "2313", "6285", "2312",
              "2330", "2317", "2382", "3231", "2308", "2368", "2449", "2454", "2303",
              "2603", "2609", "2615", "2645",
              "6257"]

for stock_id in stock_list:
    df = api.taiwan_stock_daily(
        stock_id=stock_id,
        start_date=start_date,
        end_date=end_date)

    original_len = len(df)
    abnormal = df[df["close"].isna() | (df["close"] <= 0)]
    if len(abnormal) > 0:
        print(f"\n{stock_id} 發現 {len(abnormal)} 筆 Close 異常：")
        display(abnormal[["date", "close"]])

    df = df[df["close"].notna() & (df["close"] > 0)].copy()
    df = df.reset_index(drop=True)

    df.to_csv(f"{stock_id}.csv", index=False)
    print(
        f"{stock_id}：原始 {original_len} 筆 → "
        f"保留 {len(df)} 筆 → "
        f"移除 {original_len - len(df)} 筆")

    df.to_csv(f"{stock_id}.csv", index=False)
```

```python
df_match = pd.read_csv("record.csv")
strategy_map = {
    "KD 黃金交叉／死亡交叉": ["KD黃金交叉", "KD死亡交叉"],
    "MACD Histogram": ["MACD_Hist由負翻正", "MACD_Hist由正翻負"],
    "MACD": ["MACD_Hist由負翻正", "MACD_Hist由正翻負"],
    "布林通道": ["BB中線突破", "BB上線跌破"],
    "SMA 黃金交叉／死亡交叉": ["SMA黃金交叉", "SMA死亡交叉"],
    "KD 超買／超賣": ["KD超買", "KD超賣"],
    "RSI 超買／超賣": ["RSI超買", "RSI超賣"]
}
name_to_id = {
    "中興電": "1513", "華城": "1519", "士電": "1503", "亞力": "1514",
    "鴻海": "2317", "緯創": "3231", "台積電": "2330", "廣達": "2382",
    "金像電": "2368", "京元電子": "2449", "聯電": "2303", "國巨": "2327",
    "華新科": "2492", "禾伸堂": "3026", "日電貿": "3090", "華邦電": "2344",
    "旺宏": "2337", "力積電": "6770", "南亞科": "2408", "晶豪科": "3006",
    "昇達科": "3491", "華通": "2313", "啟碁": "6285", "金寶": "2312",
    "長榮": "2603", "萬海": "2615", "陽明": "2609", "長榮航太": "2645",
    "矽格": "6257"
}
df_match["標的"] = df_match["標的"].astype(str).str.strip()
df_match["策略"] = df_match["策略"].astype(str).str.strip()

stock_strategy = {}
for _, row in df_match.iterrows():
    name = row["標的"]
    stock_id = next((code for key, code in name_to_id.items() if key in name), None)
    if stock_id:
        stock_strategy[stock_id] = strategy_map.get(row["策略"], [])

os.makedirs("Image", exist_ok=True)
os.makedirs("Result", exist_ok=True)
signal_summary = []

for stock_id in stock_list:

    print("=" * 80)
    print(f"{stock_id}")
    print("=" * 80)

    # 讀取資料
    df = pd.read_csv(f"{stock_id}.csv")
    df["date"] = pd.to_datetime(df["date"])
    df = df.sort_values("date").reset_index(drop=True)

    # KD
    df["K"], df["D"] = talib.STOCH(
        high=df["max"],
        low=df["min"],
        close=df["close"],
        fastk_period=9,
        slowk_period=3,
        slowk_matype=0,
        slowd_period=3,
        slowd_matype=0)

    # MACD
    df["MACD"], df["MACD_Signal"], df["MACD_Hist"] = talib.MACD(
        df["close"],
        fastperiod=12,
        slowperiod=26,
        signalperiod=9)

    # Bollinger Bands
    df["BB_Upper"], df["BB_Middle"], df["BB_Lower"] = talib.BBANDS(
        df["close"],
        timeperiod=20,
        nbdevup=2,
        nbdevdn=2,
        matype=0)

    # RSI
    df["RSI_7"] = talib.RSI(df["close"], timeperiod=7)
    df["RSI_14"] = talib.RSI(df["close"], timeperiod=14)

    # SMA
    df["SMA5"] = talib.SMA(df["close"], timeperiod=5)
    df["SMA10"] = talib.SMA(df["close"], timeperiod=10)
    df["SMA20"] = talib.SMA(df["close"], timeperiod=20)

    df = df.tail(60).reset_index(drop=True)

    # 畫圖
    fig, (ax1, ax2, ax3, ax4) = plt.subplots(
        4, 1,
        figsize=(14, 10),
        sharex=True,
        gridspec_kw={"height_ratios":[3, 1, 1, 1]})

    # 股價 + Bollinger Bands
    ax1.plot(df["date"], df["close"], label="Close")
    ax1.plot(df["date"], df["BB_Upper"], label="BB Upper")
    ax1.plot(df["date"], df["BB_Middle"], label="BB Middle")
    ax1.plot(df["date"], df["BB_Lower"], label="BB Lower")
    ax1.set_title(f"{stock_id} Price")
    ax1.legend()
    ax1.grid(True)

    # MACD
    ax2.plot(df["date"], df["MACD"], label="MACD")
    ax2.plot(df["date"], df["MACD_Signal"], label="Signal")
    colors = ["red" if x >= 0 else "green" for x in df["MACD_Hist"]]
    ax2.bar(df["date"], df["MACD_Hist"], color=colors, width=1)
    ax2.axhline(0, color="black")
    ax2.set_ylabel("MACD")
    ax2.legend()
    ax2.grid(True)

    # KD
    ax3.plot(df["date"], df["K"], label="K")
    ax3.plot(df["date"], df["D"], label="D")
    ax3.axhline(80, linestyle="--")
    ax3.axhline(30, linestyle="--")
    ax3.set_ylim(0,100)
    ax3.set_ylabel("KD")
    ax3.set_xlabel("Date")
    ax3.legend()
    ax3.grid(True)

    # RSI
    ax4.plot(df["date"], df["RSI_7"], label="RSI 7")
    ax4.plot( df["date"], df["RSI_14"], label="RSI 14")
    ax4.axhline(70, linestyle="--")
    ax4.axhline(30, linestyle="--")
    ax4.set_ylim(0, 100)
    ax4.set_ylabel("RSI")
    ax4.set_xlabel("Date")
    ax4.legend()
    ax4.grid(True)

    plt.tight_layout()

    # 儲存圖片
    plt.savefig(
        f"Image/{stock_id}_technical_analysis.png",
        dpi=300,
        bbox_inches="tight")

    plt.close()

    result_df = df[["date", "open", "max", "min", "close"]].copy()

    # KD
    result_df["KD黃金交叉"] = ((df["K"].shift(1) <= df["D"].shift(1)) & (df["K"] > df["D"])).astype(int)
    result_df["KD死亡交叉"] = ((df["K"].shift(1) >= df["D"].shift(1)) & (df["K"] < df["D"])).astype(int)
    result_df["KD超買"] = ((df["K"] > 80) & (df["D"] > 80)).astype(int)
    result_df["KD超賣"] = ((df["K"] < 30) & (df["D"] < 30)).astype(int)

    # MACD Histogram
    result_df["MACD_Hist由負翻正"] = ((df["MACD_Hist"].shift(1) <= 0) & (df["MACD_Hist"] > 0)).astype(int)
    result_df["MACD_Hist由正翻負"] = ((df["MACD_Hist"].shift(1) >= 0) & (df["MACD_Hist"] < 0)).astype(int)

    # Bollinger Bands
    result_df["BB中線突破"] = ((df["close"].shift(1) <= df["BB_Middle"].shift(1)) & (df["close"] > df["BB_Middle"])).astype(int)
    result_df["BB上線跌破"] = ((df["close"].shift(1) >= df["BB_Upper"].shift(1)) & (df["close"] < df["BB_Upper"])).astype(int)

    # RSI
    result_df["RSI黃金交叉"] = ((df["RSI_7"].shift(1) <= df["RSI_14"].shift(1)) & (df["RSI_7"] > df["RSI_14"])).astype(int)
    result_df["RSI死亡交叉"] = ((df["RSI_7"].shift(1) >= df["RSI_14"].shift(1)) & (df["RSI_7"] < df["RSI_14"])).astype(int)
    result_df["RSI超買"] = ( (df["RSI_7"] > 70) & (df["RSI_14"] > 70)).astype(int)
    result_df["RSI超賣"] = ((df["RSI_7"] < 30) & (df["RSI_14"] < 30)).astype(int)

    # SMA
    result_df["SMA黃金交叉"] = ((df["SMA5"].shift(1) <= df["SMA20"].shift(1)) & (df["SMA5"] > df["SMA20"])).astype(int)
    result_df["SMA死亡交叉"] = ((df["SMA5"].shift(1) >= df["SMA20"].shift(1)) & (df["SMA5"] < df["SMA20"])).astype(int)

    result_df["K"] = df["K"]
    result_df["D"] = df["D"]
    result_df["MACD"] = df["MACD"]
    result_df["MACD_Signal"] = df["MACD_Signal"]
    result_df["MACD_Hist"] = df["MACD_Hist"]
    result_df["BB_Upper"] = df["BB_Upper"]
    result_df["BB_Middle"] = df["BB_Middle"]
    result_df["BB_Lower"] = df["BB_Lower"]
    result_df["RSI_7"] = df["RSI_7"]
    result_df["RSI_14"] = df["RSI_14"]
    result_df["SMA5"] = df["SMA5"]
    result_df["SMA10"] = df["SMA10"]

    result_df.to_csv(f"Result/{stock_id}_result.csv", index=False, encoding="utf-8-sig")

    # 最近幾天觸發訊號
    signal_cols = [
    "KD黃金交叉",
    "KD死亡交叉",
    "KD超買",
    "KD超賣",
    "MACD_Hist由負翻正",
    "MACD_Hist由正翻負",
    "BB中線突破",
    "BB上線跌破",
    "RSI黃金交叉",
    "RSI死亡交叉",
    "RSI超買",
    "RSI超賣",
    "SMA黃金交叉",
    "SMA死亡交叉"]

    last1 = result_df.tail(3)
    allowed_signals = stock_strategy.get(stock_id, [])
    for _, row in last1.iterrows():
      for signal in allowed_signals:
        if signal in result_df.columns and row[signal] == 1:
          signal_summary.append({
              "日期": row["date"].strftime("%Y-%m-%d"),
              "股票代號": stock_id,
              "訊號": signal,
              "收盤價": row["close"]
          })

print("\n" + "=" * 80)
print("最近幾天技術訊號整理")

summary_df = pd.DataFrame(signal_summary)
if summary_df.empty:
    print("最近幾天沒有適合策略觸發訊號")
else:
    summary_df = summary_df.sort_values(["股票代號", "日期"]).reset_index(drop=True)
    print(summary_df.to_string(index=False))

summary_df.to_csv("Result/LastDays_Signal_Summary.csv", index=False, encoding="utf-8-sig")
```
